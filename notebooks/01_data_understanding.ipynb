```python
plt.figure(figsize=(10, 5))

plt.hist(df["GrossApproval"], bins=100)

plt.xlabel("Gross Approval Amount")
plt.ylabel("Number of Loans")
plt.title("Distribution of Loan Approval Amount")

plt.tight_layout()
plt.show()
```

```python
df.groupby("Default")["GrossApproval"].agg(
    ["count", "mean", "median", "std", "min", "max"]
)
```

```python
bins = [0, 25000, 50000, 100000, 250000, 500000, 1000000, float("inf")]

labels = [
    "<25K",
    "25K-50K",
    "50K-100K",
    "100K-250K",
    "250K-500K",
    "500K-1M",
    ">1M"
]

df["LoanAmountGroup"] = pd.cut(
    df["GrossApproval"],
    bins=bins,
    labels=labels
)

loan_default_rate = (
    df.groupby("LoanAmountGroup", observed=True)["Default"]
      .mean()
      .mul(100)
)

loan_default_rate
```

```python
loan_default_rate.plot(kind="bar", figsize=(10,5))

plt.xlabel("Loan Amount Group")
plt.ylabel("Default Rate (%)")
plt.title("Default Rate by Loan Amount")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
